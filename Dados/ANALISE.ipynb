```python
# Importação de bibliotecas
import pandas as pd 
```

```python
# Lendo arquivos no excel
# df = Data Frame

df = pd.read_excel(r"Planilhas\DataBase_v1 (1).xlsx",
    sheet_name="EQUIPE",
    usecols=["SENIORIDADE", "DESCRICAO_AT", "AUSENCIA_TIPO"],
)

df.head()
df
```

```python
# DFS disponíveis no ambiente
%whos DataFrame
```

```python
#Verificando os tipos de senioridade
print(df["SENIORIDADE"].unique())
```

```python
# Filtrando os consultores do tipo Jr
consultor_Jr = df[df["SENIORIDADE"] == "Jr"]
consultor_Jr.head(100)
```

```python
# Filtrando os consultores que estão ausentes por férias
consultores_Ausentes = df[df["AUSENCIA_TIPO"] == "FÉRIAS"]
consultores_Ausentes.head(60)
```

```python
# Verificando quais colunas existem no DataFrame
print(consultores_Ausentes.columns.tolist())
```

```python
# Filtrando somente os consultores do tipo Jr que estão ausentes por férias
consultores_Ausentes_Jr = df[(df["SENIORIDADE"] == "Jr") & (df["AUSENCIA_TIPO"] == "FÉRIAS")]
consultores_Ausentes_Jr.head(60)
```

```python
# Contagem de quantos consultores do tipo Jr estão ausentes por férias
consultores_Ausentes_Jr.count()
```

```python
# Filtrando consultores do tipo Sr  estão ausentes por férias
consultores_Ausentes_Sr = df[(df["SENIORIDADE"] == "Sr") & (df["AUSENCIA_TIPO"] == "FÉRIAS")]
consultores_Ausentes_Sr.head(100)
```

```python
# Contagem de quantos consultores do tipo Sr estão ausentes por férias
consultores_Ausentes_Sr.count()
```

```python
# Fazendo percentual relativo 
# Use len() para contar linhas corretamente
percentual_representativo = (len(consultores_Ausentes) / len(consultores_Ausentes_Sr)) * 100
print(f"{percentual_representativo:.2f}%")
```
