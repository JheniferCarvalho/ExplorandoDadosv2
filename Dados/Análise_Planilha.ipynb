```python
import pandas as pd

demandas = pd.read_excel(r"C:\Users\Jhenifer Carvalho\Desktop\ExplorandoDadosv2\Planilhas\DataBase_v1 (1).xlsx",
    sheet_name="DEMANDAS")

equipe = pd.read_excel(r"C:\Users\Jhenifer Carvalho\Desktop\ExplorandoDadosv2\Planilhas\DataBase_v1 (1).xlsx",
     sheet_name="EQUIPE")


print("DEMANDAS:", demandas.shape)
print("EQUIPE:", equipe.shape)
```

```python
df_merged = demandas.merge(
    equipe,
    on="IS",        
    how="left"      
)

print(df_merged.shape)
df_merged.head(100)
```

```python
df_merged.groupby("CONSULTOR")["CHAMADO"].count()
```

```python
df_merged.groupby("TIPO_CHAMADO")["CHAMADO"].count()
```

```python
chamados_N1 = df_merged[df_merged["COMPLEXIDADE"] == "N1"]
chamados_N1
```

```python
chamados_N1 = chamados_N1["CHAMADO"].count()
print(f"Número de chamados de complexidade N1: {chamados_N1}")
```
